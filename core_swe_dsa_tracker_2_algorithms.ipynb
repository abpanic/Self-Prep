**Approach:**

- Code walkthrough:
- For each index `i`, saves `nums[i]` as `value`, then shifts all larger elements one position right by copying `nums[j]` to `nums[j+1]`.
- Inserts `value` at the gap left by the shifts, keeping the slice `nums[:i+1]` sorted after every outer iteration.
- Prints `[1, 2, 4, 5]`.

**Trade-offs:**
- Sorting exposes structure and simplifies many scans, but costs O(n log n) and may allocate extra memory.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**

```python
# Insertion sort is useful for learning local swaps.
nums = [5, 2, 4, 1]
for i in range(1, len(nums)):
    value = nums[i]
    j = i - 1
    while j >= 0 and nums[j] > value:
        nums[j + 1] = nums[j]
        j -= 1
    nums[j + 1] = value
print(nums)
```

### Checklist item: Merge sort

**Approach:**
- First, recognize the pattern: Recursively split the array in half until single elements remain, then merge adjacent sorted halves back into a sorted result.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Merge sort recursively splits, then merges sorted halves.
- Code walkthrough:
- Splits `nums` at `mid` into two halves, recursively sorts each, then merges by comparing the front elements and appending the smaller.
- After both inner lists are exhausted or one runs out, appends the remaining tail with slice concatenation.
- Prints `[1, 2, 4, 5]`.

**Trade-offs:**
- Sorting exposes structure and simplifies many scans, but costs O(n log n) and may allocate extra memory.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for reports, leaderboards, deduping/grouping pipelines, interval merging, scheduling, and deterministic output.
- When not to use it: Do not sort huge data just to find one item when a heap, index, hash map, streaming algorithm, or database ORDER BY/LIMIT is better.


```python
# Merge sort recursively splits, then merges sorted halves.
def merge_sort(nums):
    if len(nums) <= 1:
        return nums
    mid = len(nums) // 2
    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    return result + left[i:] + right[j:]
print(merge_sort([5, 2, 4, 1]))
```

### Checklist item: Quick sort

**Approach:**
- First, recognize the pattern: Choose the ordering key, sort or partition the input, then scan/merge using the sorted invariant.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Simple quicksort for learning partitioning.
- Code walkthrough:
- Picks the middle element as `pivot`, then partitions the list into three groups: less than, equal to, and greater than `pivot`.
- Recursively sorts `left` and `right`, then concatenates all three parts; base case returns a list of length ≤ 1 unchanged.
- Prints `[1, 2, 4, 5]`.

**Trade-offs:**
- Sorting exposes structure and simplifies many scans, but costs O(n log n) and may allocate extra memory.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for reports, leaderboards, deduping/grouping pipelines, interval merging, scheduling, and deterministic output.
- When not to use it: Do not sort huge data just to find one item when a heap, index, hash map, streaming algorithm, or database ORDER BY/LIMIT is better.


```python
# Simple quicksort for learning partitioning.
def quicksort(nums):
    if len(nums) <= 1:
        return nums
    pivot = nums[len(nums) // 2]
    left = [x for x in nums if x < pivot]
    mid = [x for x in nums if x == pivot]
    right = [x for x in nums if x > pivot]
    return quicksort(left) + mid + quicksort(right)
print(quicksort([5, 2, 4, 1]))
```

### Checklist item: Classic binary search

**Approach:**
- First, recognize the pattern: Define the search interval and the monotonic condition, then update exactly one boundary per iteration.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Classic binary search.
- Code walkthrough:
- Computes `mid` as the integer midpoint of `[left, right]`, then narrows: sets `left = mid + 1` when `nums[mid]` is too small, `right = mid - 1` when too large.
- Prints `mid` as soon as `nums[mid] == target`, demonstrating that the O(log n) search finds the answer in at most log₂(n) iterations.
- Prints `3` — the index of `7` in `[1, 3, 5, 7, 9]`, found in two comparisons.

**Trade-offs:**
- Very fast search, but correctness depends on a precise invariant and boundary handling.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for sorted collections, feature threshold tuning, capacity limits, pagination bounds, and monotonic feasibility checks.

```python
# Classic binary search.
nums = [1, 3, 5, 7, 9]
target = 7
left, right = 0, len(nums) - 1
while left <= right:
    mid = (left + right) // 2
    if nums[mid] == target:
        print(mid)
        break
    if nums[mid] < target:
        left = mid + 1
    else:
        right = mid - 1
```

### Checklist item: Search in rotated sorted array

**Approach:**

**Trade-offs:**

```python
# Search in rotated sorted array.
nums = [4, 5, 6, 7, 0, 1, 2]
target = 0
left, right = 0, len(nums) - 1
while left <= right:
    mid = (left + right) // 2
    if nums[mid] == target:
        print(mid)
        break
    if nums[left] <= nums[mid]:
        if nums[left] <= target < nums[mid]:
            right = mid - 1
        else:
            left = mid + 1
    else:
        if nums[mid] < target <= nums[right]:
            left = mid + 1
        else:
            right = mid - 1
```

### Beginner-friendly intro
Recursion is when a function calls itself on a smaller part of the problem until a simple base case.
It naturally matches problems that have a self-similar or tree-like structure.
Most recursion questions are about correctly defining the base case and the step that reduces the problem size.


### Checklist item: Base case

**Approach:**
- First, recognize the pattern: Write the base case first, make each recursive call smaller, and identify what the function returns to its caller.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Recursion needs a base case and a smaller recursive case.
- Code walkthrough:

- Every other call multiplies `n` by `factorial(n - 1)`, reducing the problem size by exactly 1 each time.
- Prints `120` = 5! after 6 total calls (one per value from 5 down to 0).

**Trade-offs:**
- Readable for self-similar problems, but stack depth and repeated subcalls can become problems.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**

### Beginner-friendly intro
Backtracking builds solutions step by step and abandons (backtracks) paths that cannot lead to a valid answer.
You try a choice, explore deeper, and then undo the choice before trying the next option.
You use backtracking for combinatorial search problems like subsets, permutations, or constraint puzzles.


### Checklist item: Decision tree

**Approach:**
- First, recognize the pattern: At each position, branch into include or exclude; recurse on the remaining elements and collect valid leaves as results.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Backtracking subsets: choose/explore/unchoose.
- Code walkthrough:
- `decide(index, path)` appends `path[:]` when `index == len(nums)`, capturing the complete subset at each leaf of the decision tree.
- Each level branches into two recursive calls: first `decide(index + 1, path)` (exclude), then `path.append` + `decide(index + 1, path)` (include) + `path.pop`.
- Prints all 8 subsets of `[1, 2, 3]`, including the empty set, without an inner `for` loop.

**Trade-offs:**
- Easy to express exhaustive search, but worst-case time is often exponential.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for constraint solving, test-case generation, configuration search, scheduling possibilities, and puzzle-like domains.
- When not to use it: Do not use raw backtracking when input size is large and there is no pruning, memoization, or constraint propagation.


```python
# Decision tree: at each index, branch into 'include' or 'exclude'.
# This produces all 2^n subsets without an inner loop.
nums = [1, 2, 3]
result = []

def decide(index, path):
    if index == len(nums):
        result.append(path[:])
        return
    decide(index + 1, path)              # branch: exclude nums[index]
    path.append(nums[index])
    decide(index + 1, path)              # branch: include nums[index]
    path.pop()

decide(0, [])
print(result)
```

- The loop picks `candidates[i]` (CHOOSE), recurses with `i + 1` (EXPLORE), then removes the last element (UNCHOOSE) to restore state before the next candidate.
- Produces all 8 subsets of `['a', 'b', 'c']` using a single mutable `path` rather than creating new lists each call.

**Trade-offs:**

```python
# Choose / explore / unchoose: mutate 'path', recurse, then undo.
candidates = ['a', 'b', 'c']
result = []
path = []

def backtrack(start):
    result.append(path[:])           # snapshot the current subset
    for i in range(start, len(candidates)):
        path.append(candidates[i])   # CHOOSE
        backtrack(i + 1)             # EXPLORE
        path.pop()                   # UNCHOOSE (restore state)

backtrack(0)
print(result)
```

### Beginner-friendly intro
A sliding window keeps track of a range (window) over an array or string while you move its start or end.
The idea is to avoid recomputing work from scratch when the window shifts.
You use it for problems that ask about contiguous subarrays or substrings with some property (length, sum, unique chars).


### Checklist item: Fixed-size window

**Approach:**
- First, recognize the pattern: Define the valid-window condition, expand with the right pointer, update state, then shrink from the left until the condition is restored.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Fixed-size sliding window: max sum of any k consecutive values.
- Code walkthrough:
- Computes the initial window sum for the first `k = 3` elements and stores it in `best`.
- Slides by adding the incoming element (`nums[right]`) and subtracting the outgoing one (`nums[right - k]`) each step.
- Prints `16` — the maximum sum window `[1, 7, 8]` in `[4, 2, 1, 7, 8, 1]`.

**Trade-offs:**
- Usually O(n), but only when each pointer advances monotonically and window state is maintained correctly.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for streaming metrics, log windows, rate limits, rolling aggregates, and substring/subarray constraints.
- When not to use it: Do not use it when the valid region is not contiguous or when elements can move backward arbitrarily.


```python
# Fixed-size sliding window: max sum of any k consecutive values.
nums = [4, 2, 1, 7, 8, 1]
k = 3
window = sum(nums[:k])
best = window
for right in range(k, len(nums)):
    window += nums[right] - nums[right - k]
    best = max(best, window)
print(best)
```

Dynamic programming breaks a problem into overlapping subproblems and stores the answers so you do not recompute them.
You usually define a dp[state] that captures a smaller subproblem and build up to the final answer.
DP shines in optimization and counting problems where brute force would revisit the same subcases many times.

**Approach:**

- Then trace a tiny example before coding. For this sample, the code demonstrates: Top-down memoization.
- Code walkthrough:
- `@lru_cache(None)` caches every `fib(n)` result so each unique `n` is computed only once across all recursive calls.
- Without the cache, `fib(30)` would make ~2³⁰ calls; with it, only 31 unique calls are made, reducing complexity from O(2^n) to O(n).
- Prints `832040` — the 30th Fibonacci number.

**Trade-offs:**

```python
# Top-down memoization.
from functools import lru_cache
@lru_cache(None)
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)
print(fib(30))
```

### Beginner-friendly intro
Depth-First Search (DFS) explores a graph by going as deep as possible along each branch before backtracking.
Breadth-First Search (BFS) explores layer by layer, visiting all neighbors at each distance before going deeper.
You use DFS and BFS to test reachability, count components, or traverse all nodes in a graph or grid.


### Checklist item: DFS recursive

**Approach:**
- First, recognize the pattern: Choose adjacency representation, initialize visited/indegree/distance state, then traverse while preserving the graph invariant.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Recursive DFS.
- Code walkthrough:
- `dfs(node)` marks the node as seen then recurses into each unvisited neighbour.
- Handles the 3-node graph `{'A': ['B', 'C'], 'B': [], 'C': []}` starting from `'A'`, visiting all reachable nodes.
- Prints `{'A', 'B', 'C'}` — the full set of visited nodes.

**Trade-offs:**
- Simple and general, but memory/stack usage can be large on wide or deep graphs.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for crawlers, dependency scans, permission reachability, connected components, grid search, and shortest unweighted paths.
- When not to use it: Do not use plain BFS/DFS for weighted shortest paths; use Dijkstra, Bellman-Ford, or another weighted algorithm.


```python
# Recursive DFS.
graph = {"A": ["B", "C"], "B": [], "C": []}
seen = set()
def dfs(node):
    seen.add(node)
    for neighbor in graph[node]:
        if neighbor not in seen:
            dfs(neighbor)
dfs("A")
print(seen)
```

```python
# Visited set on a grid: mark cells before queuing to avoid revisits.
from collections import deque
grid = [
    [1, 1, 0],
    [0, 1, 1],
    [0, 0, 1],
]
rows, cols = len(grid), len(grid[0])
visited = set()

def bfs(r, c):
    queue = deque([(r, c)])
    visited.add((r, c))
    while queue:
        row, col = queue.popleft()
        for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nr, nc = row + dr, col + dc
            if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 1 and (nr, nc) not in visited:
                visited.add((nr, nc))
                queue.append((nr, nc))

bfs(0, 0)
print('reachable 1-cells from (0,0):', sorted(visited))
```

### Beginner-friendly intro
Shortest path algorithms find the cheapest way to travel from one node to another in a graph based on edge weights.
For unweighted graphs, BFS gives shortest paths by number of steps; for weighted graphs with non-negative edge weights, Dijkstra uses a priority queue to pick the next best node.
These algorithms are used in routing, navigation, and any place you want minimum cost between points.


### Checklist item: BFS for unweighted shortest path

**Approach:**
- First, recognize the pattern: Put starting states into the queue, mark them seen, then process one layer/frontier at a time.
- Then trace a tiny example before coding. For this sample, the code demonstrates: BFS shortest path in an unweighted graph.
- Code walkthrough:
- Queues tuples `(node, dist)`, starting with `('A', 0)`; each dequeued node's unseen neighbours are enqueued with `dist + 1`.
- Stops and prints `dist` as soon as the target `'D'` is dequeued — BFS guarantees this is the shortest path.
- Prints `2` — the shortest path from `'A'` to `'D'` is `A→B→D` or `A→C→D`, each 2 hops.

**Trade-offs:**

```python
# BFS shortest path in an unweighted graph.
from collections import deque
graph = {"A": ["B", "C"], "B": ["D"], "C": ["D"], "D": []}
queue = deque([("A", 0)])
seen = {"A"}
while queue:
    node, dist = queue.popleft()
    if node == "D":
        print(dist)
        break
    for neighbor in graph[node]:
        if neighbor not in seen:
            seen.add(neighbor)
            queue.append((neighbor, dist + 1))
```

### Checklist item: Dijkstra's algorithm for non-negative weighted graphs

**Approach:**
- First, recognize the pattern: Initialize the source distance to zero, repeatedly expand the cheapest known node from a priority queue, and relax outgoing edges when you find a cheaper path.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Dijkstra for non-negative edge weights.
- Code walkthrough:
- Initialises `dist = {'A': 0}` and a min-heap with `(0, 'A')`; pops the lowest-cost state each iteration.
- Skips a popped entry if its `cost` differs from `dist[node]` — this filters out stale heap entries after a shorter path was found.
- Prints `{'A': 0, 'B': 2, 'C': 3}` — the cheapest path to `C` is `A→B→C` (cost 2+1=3), not `A→C` (cost 5).

**Trade-offs:**
- Precise minimum-cost paths, but the right algorithm depends on edge weights and graph size.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for routing, dependency costs, workflow transition costs, network latency paths, and least-cost planning.
- When not to use it: Do not use Dijkstra with negative edge weights, and do not use weighted algorithms when unweighted BFS is enough.


```python
# Dijkstra for non-negative edge weights.
import heapq
graph = {"A": [("B", 2), ("C", 5)], "B": [("C", 1)], "C": []}
dist = {"A": 0}
heap = [(0, "A")]
while heap:
    cost, node = heapq.heappop(heap)
    if cost != dist[node]:
        continue
    for neighbor, weight in graph[node]:
        new_cost = cost + weight
        if new_cost < dist.get(neighbor, float("inf")):
            dist[neighbor] = new_cost
            heapq.heappush(heap, (new_cost, neighbor))
print(dist)
```

- Code walkthrough:
- Initialises all distances to infinity except source `'A'` at `0`, then relaxes every edge `len(dist) - 1` times.
- Each pass updates `dist[b]` whenever `dist[a] + weight < dist[b]`, propagating improvements across the graph.
- Prints `{'A': 0, 'B': 4, 'C': 2}` — the negative edge `B→C(−2)` makes `A→B→C = 2` cheaper than `A→C = 5`.

**Trade-offs:**

```python
# Bellman-Ford supports negative edges and detects negative cycles.
edges = [("A", "B", 4), ("A", "C", 5), ("B", "C", -2)]
dist = {"A": 0, "B": float("inf"), "C": float("inf")}
for _ in range(len(dist) - 1):
    for a, b, weight in edges:
        if dist[a] + weight < dist[b]:
            dist[b] = dist[a] + weight
print(dist)
```

### Checklist item: Indegree method / Kahn's algorithm

**Approach:**
- First, recognize the pattern: Compute each node's indegree, enqueue zero-indegree nodes, pop one per step and decrement neighbors — enqueue any that reach zero.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Kahn's algorithm for topological sort.
- Code walkthrough:
- Computes the indegree of every node, then seeds the queue with all zero-indegree nodes (no prerequisites needed).
- Each dequeued node goes into `order`, and its neighbours' indegrees are decremented; those reaching `0` join the queue.
- Prints `['shop', 'cook', 'eat']` — a valid topological order for this dependency graph.

**Trade-offs:**

```python
# Kahn's algorithm for topological sort.
from collections import deque
graph = {"shop": ["cook"], "cook": ["eat"], "eat": []}
indegree = {node: 0 for node in graph}
for node in graph:
    for neighbor in graph[node]:
        indegree[neighbor] += 1
queue = deque([node for node, degree in indegree.items() if degree == 0])
order = []
while queue:
    node = queue.popleft()
    order.append(node)
    for neighbor in graph[node]:
        indegree[neighbor] -= 1
        if indegree[neighbor] == 0:
            queue.append(neighbor)
print(order)
```

```python
# DFS topological sort appends after visiting descendants.
graph = {"shop": ["cook"], "cook": ["eat"], "eat": []}
seen = set()
order = []
def dfs(node):
    seen.add(node)
    for neighbor in graph[node]:
        if neighbor not in seen:
            dfs(neighbor)
    order.append(node)
for node in graph:
    if node not in seen:
        dfs(node)
print(list(reversed(order)))
```
