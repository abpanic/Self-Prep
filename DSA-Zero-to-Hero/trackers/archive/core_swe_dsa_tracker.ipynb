### Beginner-friendly intro
Arrays are ordered lists of items stored one after another in memory, and strings are just arrays of characters.
You use them when you care about positions (index 0, index 1, etc.) and want fast reads and writes by index.
Most interview problems here are about scanning, slicing, and combining values while keeping track of indices and ranges.


### Checklist item: Array traversal, indexing, slicing, and in-place updates

**Approach:**
- First, recognize the pattern: Decide whether you only need to read values or also mutate them. If you mutate, loop with indexes so each write is intentional and traceable.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Array traversal, indexing, slicing, and in-place updates.
- Code walkthrough:
- Uses `enumerate` to access both index and value in one loop over `nums = [3, 1, 4, 1, 5]`.
- Accumulates `running_total` from the original values before any mutation.
- Doubles odd values in-place with `nums[index] = value * 2`, showing safe index-based mutation without creating a new list.

**Trade-offs:**
- Fast and memory-friendly for sequential work, but index errors and accidental in-place mutation can create subtle production bugs.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this in services that process ordered data: request logs, CSV rows, byte buffers, validation strings, pagination windows, and UI lists.

### Checklist item: Prefix sums and difference arrays

**Approach:**
- First, recognize the pattern: Precompute cumulative state so each range query or batch update avoids rescanning the original array.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Prefix sums answer repeated range-sum queries in O(1) after O(n) setup.
- Code walkthrough:
- Builds `prefix` starting with a sentinel `0` so every range maps cleanly: `prefix[i+1]` holds the sum of `nums[:i+1]`.
- Defines `range_sum(left, right)` as a single subtraction: `prefix[right + 1] - prefix[left]`, giving O(1) per query after O(n) setup.
- Calls `range_sum(1, 3)` to return the sum of indices 1–3 (values 4+1+7 = 12).

**Trade-offs:**
- Fast and memory-friendly for sequential work, but index errors and accidental in-place mutation can create subtle production bugs.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this in services that process ordered data: request logs, CSV rows, byte buffers, validation strings, pagination windows, and UI lists.
- When not to use it: Do not use manual index-heavy code when a clearer library call, parser, database query, or vectorized operation already expresses the same intent safely.


```python
# Prefix sums answer repeated range-sum queries in O(1) after O(n) setup.
nums = [2, 4, 1, 7, 3]
prefix = [0]
for value in nums:
    prefix.append(prefix[-1] + value)

def range_sum(left, right):
    return prefix[right + 1] - prefix[left]

print(range_sum(1, 3))
```

- Advances `left` when the sum is below `target`; retreats `right` when above — guaranteed to find the pair in a sorted array without revisiting any index.
- Prints the index pair `(1, 3)` when `nums[1] + nums[3] == 9`, then breaks immediately.

**Trade-offs:**
- Fast and memory-friendly for sequential work, but index errors and accidental in-place mutation can create subtle production bugs.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**

```python
# Two pointers on a sorted array.
nums = [1, 2, 4, 7, 11]
target = 9
left, right = 0, len(nums) - 1
while left < right:
    total = nums[left] + nums[right]
    if total == target:
        print((left, right))
        break
    if total < target:
        left += 1
    else:
        right -= 1
```

### Checklist item: Sliding window on arrays and strings

**Approach:**
- First, recognize the pattern: Define the valid-window condition, expand with the right pointer, update state, then shrink from the left until the condition is restored.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Sliding window: max sum of any k consecutive values.
- Code walkthrough:
- Computes the initial window sum for the first `k = 3` elements with `sum(nums[:k])` and stores it in `best`.
- Slides one step at a time: adds the incoming right element (`nums[right]`) and subtracts the outgoing left element (`nums[right - k]`).
- Keeps `best` as the running maximum, so only one full scan is needed after initialisation.

**Trade-offs:**
- Fast and memory-friendly for sequential work, but index errors and accidental in-place mutation can create subtle production bugs.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this in services that process ordered data: request logs, CSV rows, byte buffers, validation strings, pagination windows, and UI lists.
- When not to use it: Do not use manual index-heavy code when a clearer library call, parser, database query, or vectorized operation already expresses the same intent safely.


```python
# Sliding window: max sum of any k consecutive values.
nums = [4, 2, 1, 7, 8, 1]
k = 3
window = sum(nums[:k])
best = window
for right in range(k, len(nums)):
    window += nums[right] - nums[right - k]
    best = max(best, window)
print(best)
```

### Beginner-friendly intro
Stacks are Last-In-First-Out: the last item you put in comes out first, like an undo history.
Queues are First-In-First-Out: the first item you put in comes out first, like people standing in a line.
You use stacks when you model nested operations or backtracking, and queues when you process items in arrival order (like BFS).


### Checklist item: Stack operations and LIFO reasoning

**Approach:**
- First, recognize the pattern: Define what belongs on the stack and when it becomes resolved, then pop only when the current item proves a previous item is complete.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Stack: last in, first out.
- Code walkthrough:
- Pushes `'open'`, `'work'`, and `'close'` with `append`, so `'close'` ends up on top.
- Drains the stack with `pop`, printing `'close'`, `'work'`, `'open'` in LIFO order — last pushed is first popped.

**Trade-offs:**
- Simple and predictable ordering, but correctness depends on adding/removing items at the right side at the right time.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this in parsers, undo/redo stacks, job queues, breadth-first crawlers, rate-limited workers, and streaming processors.
- When not to use it: Do not use a stack or queue when random access, priority ordering, or durable cross-process delivery is the real requirement.


```python
# Stack: last in, first out.
stack = []
for token in ["open", "work", "close"]:
    stack.append(token)
while stack:
    print(stack.pop())
```

### Checklist item: Valid parentheses

**Approach:**
- First, recognize the pattern: Define what belongs on the stack and when it becomes resolved, then pop only when the current item proves a previous item is complete.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Validate balanced parentheses with a stack.
- Code walkthrough:
- Builds a `pairs` dict mapping each closer to its expected opener.
- Pushes every opener onto `stack`; on a closer, pops and checks whether the top matches — if not, returns `False` immediately.
- Returns `not stack` at the end so any unclosed opener also causes a failure.

**Trade-offs:**
- Simple and predictable ordering, but correctness depends on adding/removing items at the right side at the right time.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this in parsers, undo/redo stacks, job queues, breadth-first crawlers, rate-limited workers, and streaming processors.
- When not to use it: Do not use a stack or queue when random access, priority ordering, or durable cross-process delivery is the real requirement.


```python
# Validate balanced parentheses with a stack.
def is_valid(s):
    pairs = {")": "(", "]": "[", "}": "{"}
    stack = []
    for ch in s:
        if ch in pairs.values():
            stack.append(ch)
        elif ch in pairs and (not stack or stack.pop() != pairs[ch]):
            return False
    return not stack

print(is_valid("{[()]}"))
```

### Beginner-friendly intro
A binary search tree is a tree where every value in the left subtree is smaller and every value in the right subtree is larger, based on the duplicate-value rule the implementation chooses.
Because of this ordering, you can search, insert, and delete while always moving left or right based on comparisons.
BST questions focus on using this ordering property to make operations faster than scanning the whole tree.


### Checklist item: BST property

**Approach:**
- First, recognize the pattern: Carry the BST ordering bounds through the operation so each comparison discards an entire subtree.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Validate BST with lower/upper bounds.
- Code walkthrough:
- Passes a narrowing `(low, high)` interval down the tree: every node must satisfy `low < node.value < high`.
- Recursing left tightens the upper bound to the parent's value; recursing right tightens the lower bound.
- Returns `True` for the valid BST `Node(5, Node(3), Node(8))`.

**Trade-offs:**
- Balanced trees give O(log n) operations, but skewed trees degrade toward O(n).
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for ordered data where predecessor, successor, min/max, or range-style navigation matters.
- When not to use it: Do not use a plain unbalanced BST for production indexing unless you can guarantee balance; use a library-backed balanced tree or database index.


```python
# Validate BST with lower/upper bounds.
class Node:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right

root = Node(5, Node(3), Node(8))
def is_bst(node, low=float("-inf"), high=float("inf")):
    if node is None:
        return True
    return low < node.value < high and is_bst(node.left, low, node.value) and is_bst(node.right, node.value, high)
print(is_bst(root))
```

### Checklist item: Search, insert, and delete

**Approach:**
- First, recognize the pattern: Start from the brute-force version, name the state that removes repeated work, then implement the smallest version that preserves that state.
- Then trace a tiny example before coding. For this sample, the code demonstrates: BST search and insert. Deletion has more cases, so practice it separately.
- Code walkthrough:
- `insert` navigates left or right based on value comparison and creates a new `Node` when it reaches `None`, returning it up the call stack.
- Inserts `5, 3, 8, 4` in sequence; `4` lands as the right child of `3` (left subtree of root `5`).
- Prints `root.left.right.value = 4`, confirming the node was placed correctly.

**Trade-offs:**

```python
# BST search and insert. Deletion has more cases, so practice it separately.
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def insert(node, value):
    if node is None:
        return Node(value)
    if value < node.value:
        node.left = insert(node.left, value)
    elif value > node.value:
        node.right = insert(node.right, value)
    return node

root = None
for value in [5, 3, 8, 4]:
    root = insert(root, value)
print(root.left.right.value)
```

```python
# LCA in a BST uses ordering to move toward the split point.
class Node:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right

root = Node(5, Node(3), Node(8))
def lca_value(root, a, b):
    node = root
    while node:
        if a < node.value and b < node.value:
            node = node.left
        elif a > node.value and b > node.value:
            node = node.right
        else:
            return node.value

print(lca_value(root, 3, 8))
```

```python
# Multi-source BFS starts all sources at distance 0.
from collections import deque
grid = [[0, 1, 0], [0, 0, 0], [1, 0, 0]]
rows, cols = len(grid), len(grid[0])
dist = [[None] * cols for _ in range(rows)]
queue = deque()
for r in range(rows):
    for c in range(cols):
        if grid[r][c] == 1:
            dist[r][c] = 0
            queue.append((r, c))
while queue:
    r, c = queue.popleft()
    for dr, dc in [(1,0), (-1,0), (0,1), (0,-1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < rows and 0 <= nc < cols and dist[nr][nc] is None:
            dist[nr][nc] = dist[r][c] + 1
            queue.append((nr, nc))
print(dist)
```

```python
# Kruskal MST sketch with Union-Find.
edges = [(1, "A", "B"), (3, "A", "C"), (2, "B", "C")]
parent = {node: node for edge in edges for node in edge[1:]}
def find(x):
    while parent[x] != x:
        parent[x] = parent[parent[x]]
        x = parent[x]
    return x
mst = []
for weight, a, b in sorted(edges):
    ra, rb = find(a), find(b)
    if ra != rb:
        parent[ra] = rb
        mst.append((weight, a, b))
print(mst)
```

```python
# Kosaraju's SCC: DFS on original graph records finish order;
# DFS on reversed graph in reverse finish order finds each SCC.
graph = {'A': ['B'], 'B': ['C'], 'C': ['A'], 'D': ['C']}
visited, finish_stack = set(), []

def dfs1(node):
    visited.add(node)
    for nb in graph[node]:
        if nb not in visited:
            dfs1(nb)
    finish_stack.append(node)

for node in graph:
    if node not in visited:
        dfs1(node)

rev = {node: [] for node in graph}
for node, neighbors in graph.items():
    for nb in neighbors:
        rev[nb].append(node)

visited2, sccs = set(), []

def dfs2(node, scc):
    visited2.add(node)
    scc.append(node)
    for nb in rev[node]:
        if nb not in visited2:
            dfs2(nb, scc)

while finish_stack:
    node = finish_stack.pop()
    if node not in visited2:
        scc = []
        dfs2(node, scc)
        sccs.append(sorted(scc))

print('SCCs:', sccs)   # [['A', 'B', 'C'], ['D']]
```

### Checklist item: Bitmask DP on graphs (small N: Hamiltonian paths / traveling salesman-style problems)

**Approach:**
- First, recognize the pattern: Compute subtree answers bottom-up, then reroot by propagating a corrected parent contribution downward in a second DFS pass.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Bitmask DP: visited set encoded as bits.
- Code walkthrough:
- Encodes visited nodes as bits: `dp[(mask, last)]` counts paths that have visited exactly the nodes in `mask` and ended at node `last`.
- Seeds with `dp[(1<<0, 0)] = 1` — one path starting at node 0 with only node 0 visited — then expands to each unvisited next node.
- Sums `dp[((1<<n)-1, last)]` over all possible last nodes to get the total number of Hamiltonian paths in the 3-node graph.

**Trade-offs:**
- Can turn repeated exponential work into a tractable computation, but state design is the hard part.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for repeated substructure in dependency DAGs, tree aggregations, recommendation paths, permissions inheritance, and small-state graph optimization.
- When not to use it: Do not use DP if subproblems do not overlap or if the state space explodes beyond memory/time limits.


```python
# Bitmask DP: visited set encoded as bits.
n = 3
dp = {(1 << 0, 0): 1}
for mask in range(1 << n):
    for last in range(n):
        ways = dp.get((mask, last), 0)
        if not ways:
            continue
        for nxt in range(n):
            if not (mask & (1 << nxt)):
                dp[(mask | (1 << nxt), nxt)] = dp.get((mask | (1 << nxt), nxt), 0) + ways
print(sum(dp.get(((1 << n) - 1, last), 0) for last in range(n)))
```

```python
# starts_with checks if any word begins with the prefix.
# words_with_prefix collects every inserted word that does.
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_word = False

root = TrieNode()

def insert(word):
    node = root
    for ch in word:
        node = node.children.setdefault(ch, TrieNode())
    node.is_word = True

def starts_with(prefix):
    node = root
    for ch in prefix:
        if ch not in node.children:
            return False
        node = node.children[ch]
    return True

def words_with_prefix(prefix):
    node = root
    for ch in prefix:
        if ch not in node.children:
            return []
        node = node.children[ch]
    results = []
    def collect(n, current):
        if n.is_word:
            results.append(current)
        for ch, child in n.children.items():
            collect(child, current + ch)
    collect(node, prefix)
    return results

for word in ['car', 'card', 'care', 'cat']:
    insert(word)
print(starts_with('car'))          # True
print(starts_with('dog'))          # False
print(words_with_prefix('car'))    # ['car', 'card', 'care']
```

### Checklist item: Merge sort

**Approach:**
- First, recognize the pattern: Recursively split the array in half until single elements remain, then merge adjacent sorted halves back into a sorted result.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Merge sort recursively splits, then merges sorted halves.
- Code walkthrough:
- Splits `nums` at `mid` into two halves, recursively sorts each, then merges by comparing the front elements and appending the smaller.
- After both inner lists are exhausted or one runs out, appends the remaining tail with slice concatenation.
- Prints `[1, 2, 4, 5]`.

**Trade-offs:**
- Sorting exposes structure and simplifies many scans, but costs O(n log n) and may allocate extra memory.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for reports, leaderboards, deduping/grouping pipelines, interval merging, scheduling, and deterministic output.
- When not to use it: Do not sort huge data just to find one item when a heap, index, hash map, streaming algorithm, or database ORDER BY/LIMIT is better.


```python
# Merge sort recursively splits, then merges sorted halves.
def merge_sort(nums):
    if len(nums) <= 1:
        return nums
    mid = len(nums) // 2
    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    return result + left[i:] + right[j:]
print(merge_sort([5, 2, 4, 1]))
```

```python
# Simple quicksort for learning partitioning.
def quicksort(nums):
    if len(nums) <= 1:
        return nums
    pivot = nums[len(nums) // 2]
    left = [x for x in nums if x < pivot]
    mid = [x for x in nums if x == pivot]
    right = [x for x in nums if x > pivot]
    return quicksort(left) + mid + quicksort(right)
print(quicksort([5, 2, 4, 1]))
```

### Beginner-friendly intro
Backtracking builds solutions step by step and abandons (backtracks) paths that cannot lead to a valid answer.
You try a choice, explore deeper, and then undo the choice before trying the next option.
You use backtracking for combinatorial search problems like subsets, permutations, or constraint puzzles.


### Checklist item: Decision tree

**Approach:**
- First, recognize the pattern: At each position, branch into include or exclude; recurse on the remaining elements and collect valid leaves as results.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Backtracking subsets: choose/explore/unchoose.
- Code walkthrough:
- `decide(index, path)` appends `path[:]` when `index == len(nums)`, capturing the complete subset at each leaf of the decision tree.
- Each level branches into two recursive calls: first `decide(index + 1, path)` (exclude), then `path.append` + `decide(index + 1, path)` (include) + `path.pop`.
- Prints all 8 subsets of `[1, 2, 3]`, including the empty set, without an inner `for` loop.

**Trade-offs:**
- Easy to express exhaustive search, but worst-case time is often exponential.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for constraint solving, test-case generation, configuration search, scheduling possibilities, and puzzle-like domains.
- When not to use it: Do not use raw backtracking when input size is large and there is no pruning, memoization, or constraint propagation.


```python
# Decision tree: at each index, branch into 'include' or 'exclude'.
# This produces all 2^n subsets without an inner loop.
nums = [1, 2, 3]
result = []

def decide(index, path):
    if index == len(nums):
        result.append(path[:])
        return
    decide(index + 1, path)              # branch: exclude nums[index]
    path.append(nums[index])
    decide(index + 1, path)              # branch: include nums[index]
    path.pop()

decide(0, [])
print(result)
```

- Code walkthrough:
- `backtrack(start)` snapshots `path[:]` at every call so every current subset (including the empty one) is recorded.
- The loop picks `candidates[i]` (CHOOSE), recurses with `i + 1` (EXPLORE), then removes the last element (UNCHOOSE) to restore state before the next candidate.
- Produces all 8 subsets of `['a', 'b', 'c']` using a single mutable `path` rather than creating new lists each call.

**Trade-offs:**

```python
# Choose / explore / unchoose: mutate 'path', recurse, then undo.
candidates = ['a', 'b', 'c']
result = []
path = []

def backtrack(start):
    result.append(path[:])           # snapshot the current subset
    for i in range(start, len(candidates)):
        path.append(candidates[i])   # CHOOSE
        backtrack(i + 1)             # EXPLORE
        path.pop()                   # UNCHOOSE (restore state)

backtrack(0)
print(result)
```

### Beginner-friendly intro
A sliding window keeps track of a range (window) over an array or string while you move its start or end.
The idea is to avoid recomputing work from scratch when the window shifts.
You use it for problems that ask about contiguous subarrays or substrings with some property (length, sum, unique chars).


### Checklist item: Fixed-size window

```python
# Fixed-size sliding window: max sum of any k consecutive values.
nums = [4, 2, 1, 7, 8, 1]
k = 3
window = sum(nums[:k])
best = window
for right in range(k, len(nums)):
    window += nums[right] - nums[right - k]
    best = max(best, window)
print(best)
```

### Checklist item: Frequency window

**Approach:**

- Code walkthrough:
- Maintains `counts`, a dict tracking how many times each character appears in the current window.
- When `counts[ch] > 1`, shrinks from `left` by decrementing `counts[s[left]]` and advancing `left` until the duplicate is gone.
- Prints `5` — the longest substring of `'abcaef'` without repeating characters is `'bcaef'`.

**Trade-offs:**
- Usually O(n), but only when each pointer advances monotonically and window state is maintained correctly.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

```python
# Frequency window: longest substring without repeated chars.
s = "abcaef"
left = 0
counts = {}
best = 0
for right, ch in enumerate(s):
    counts[ch] = counts.get(ch, 0) + 1
    while counts[ch] > 1:
        counts[s[left]] -= 1
        left += 1
    best = max(best, right - left + 1)
print(best)
```

Dynamic programming breaks a problem into overlapping subproblems and stores the answers so you do not recompute them.

**Approach:**

- Code walkthrough:
- `@lru_cache(None)` caches every `fib(n)` result so each unique `n` is computed only once across all recursive calls.
- Without the cache, `fib(30)` would make ~2³⁰ calls; with it, only 31 unique calls are made, reducing complexity from O(2^n) to O(n).

**Trade-offs:**

```python
# Top-down memoization.
from functools import lru_cache
@lru_cache(None)
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)
print(fib(30))
```

```python
# Visited set on a grid: mark cells before queuing to avoid revisits.
from collections import deque
grid = [
    [1, 1, 0],
    [0, 1, 1],
    [0, 0, 1],
]
rows, cols = len(grid), len(grid[0])
visited = set()

def bfs(r, c):
    queue = deque([(r, c)])
    visited.add((r, c))
    while queue:
        row, col = queue.popleft()
        for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nr, nc = row + dr, col + dc
            if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 1 and (nr, nc) not in visited:
                visited.add((nr, nc))
                queue.append((nr, nc))

bfs(0, 0)
print('reachable 1-cells from (0,0):', sorted(visited))
```

### Beginner-friendly intro
Shortest path algorithms find the cheapest way to travel from one node to another in a graph based on edge weights.
For unweighted graphs, BFS gives shortest paths by number of steps; for weighted graphs with non-negative edge weights, Dijkstra uses a priority queue to pick the next best node.
These algorithms are used in routing, navigation, and any place you want minimum cost between points.

**Approach:**
- First, recognize the pattern: Put starting states into the queue, mark them seen, then process one layer/frontier at a time.

- Code walkthrough:
- Queues tuples `(node, dist)`, starting with `('A', 0)`; each dequeued node's unseen neighbours are enqueued with `dist + 1`.
- Stops and prints `dist` as soon as the target `'D'` is dequeued — BFS guarantees this is the shortest path.
- Prints `2` — the shortest path from `'A'` to `'D'` is `A→B→D` or `A→C→D`, each 2 hops.

**Trade-offs:**

```python
# BFS shortest path in an unweighted graph.
from collections import deque
graph = {"A": ["B", "C"], "B": ["D"], "C": ["D"], "D": []}
queue = deque([("A", 0)])
seen = {"A"}
while queue:
    node, dist = queue.popleft()
    if node == "D":
        print(dist)
        break
    for neighbor in graph[node]:
        if neighbor not in seen:
            seen.add(neighbor)
            queue.append((neighbor, dist + 1))
```

### Checklist item: Dijkstra's algorithm for non-negative weighted graphs

**Approach:**
- First, recognize the pattern: Initialize the source distance to zero, repeatedly expand the cheapest known node from a priority queue, and relax outgoing edges when you find a cheaper path.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Dijkstra for non-negative edge weights.
- Code walkthrough:
- Initialises `dist = {'A': 0}` and a min-heap with `(0, 'A')`; pops the lowest-cost state each iteration.
- Skips a popped entry if its `cost` differs from `dist[node]` — this filters out stale heap entries after a shorter path was found.
- Prints `{'A': 0, 'B': 2, 'C': 3}` — the cheapest path to `C` is `A→B→C` (cost 2+1=3), not `A→C` (cost 5).

**Trade-offs:**
- Precise minimum-cost paths, but the right algorithm depends on edge weights and graph size.
- In production code, prefer the simplest implementation that meets the latency, memory, and maintainability requirements. Add tests for empty input, single-item input, duplicates, and boundary values.

**Practical software engineering use cases:**
- When to use it: Use this for routing, dependency costs, workflow transition costs, network latency paths, and least-cost planning.
- When not to use it: Do not use Dijkstra with negative edge weights, and do not use weighted algorithms when unweighted BFS is enough.


```python
# Dijkstra for non-negative edge weights.
import heapq
graph = {"A": [("B", 2), ("C", 5)], "B": [("C", 1)], "C": []}
dist = {"A": 0}
heap = [(0, "A")]
while heap:
    cost, node = heapq.heappop(heap)
    if cost != dist[node]:
        continue
    for neighbor, weight in graph[node]:
        new_cost = cost + weight
        if new_cost < dist.get(neighbor, float("inf")):
            dist[neighbor] = new_cost
            heapq.heappush(heap, (new_cost, neighbor))
print(dist)
```

### Checklist item: Bellman-Ford concept

- Each pass updates `dist[b]` whenever `dist[a] + weight < dist[b]`, propagating improvements across the graph.
- Prints `{'A': 0, 'B': 4, 'C': 2}` — the negative edge `B→C(−2)` makes `A→B→C = 2` cheaper than `A→C = 5`.

**Trade-offs:**

```python
# Bellman-Ford supports negative edges and detects negative cycles.
edges = [("A", "B", 4), ("A", "C", 5), ("B", "C", -2)]
dist = {"A": 0, "B": float("inf"), "C": float("inf")}
for _ in range(len(dist) - 1):
    for a, b, weight in edges:
        if dist[a] + weight < dist[b]:
            dist[b] = dist[a] + weight
print(dist)
```

### Checklist item: Indegree method / Kahn's algorithm

**Approach:**
- First, recognize the pattern: Compute each node's indegree, enqueue zero-indegree nodes, pop one per step and decrement neighbors — enqueue any that reach zero.
- Then trace a tiny example before coding. For this sample, the code demonstrates: Kahn's algorithm for topological sort.
- Code walkthrough:
- Computes the indegree of every node, then seeds the queue with all zero-indegree nodes (no prerequisites needed).
- Each dequeued node goes into `order`, and its neighbours' indegrees are decremented; those reaching `0` join the queue.
- Prints `['shop', 'cook', 'eat']` — a valid topological order for this dependency graph.

**Trade-offs:**
- Produces a dependency-safe order, but only for directed acyclic graphs.

```python
# Kahn's algorithm for topological sort.
from collections import deque
graph = {"shop": ["cook"], "cook": ["eat"], "eat": []}
indegree = {node: 0 for node in graph}
for node in graph:
    for neighbor in graph[node]:
        indegree[neighbor] += 1
queue = deque([node for node, degree in indegree.items() if degree == 0])
order = []
while queue:
    node = queue.popleft()
    order.append(node)
    for neighbor in graph[node]:
        indegree[neighbor] -= 1
        if indegree[neighbor] == 0:
            queue.append(neighbor)
print(order)
```

- Then trace a tiny example before coding. For this sample, the code demonstrates: XOR cancels duplicate pairs.
- Code walkthrough:
- XOR is self-inverse (`x ^ x == 0`) and identity-preserving (`x ^ 0 == x`), so pairing duplicates cancel out.
- Folding all values with `^=` eliminates every number appearing twice, leaving only the unpaired element.
- Prints `4` — the single number in `[4, 1, 2, 1, 2]`.

**Trade-offs:**
- Fast and compact, but can become unreadable without clear constants and tests.

```python
# XOR cancels duplicate pairs.
nums = [4, 1, 2, 1, 2]
answer = 0
for value in nums:
    answer ^= value
print(answer)
```

- Code walkthrough:
- Defines three permission flags as distinct powers of 2 by left-shifting: `READ=1`, `WRITE=2`, `EXECUTE=4`.
- Combines flags with `|`, checks presence with `& READ`, adds a flag with `|=`, and clears a flag with `&= ~WRITE`.
- Prints `True` (READ is set initially), then `0b101` (READ | EXECUTE after WRITE is cleared).

**Trade-offs:**

```python
# Bit mask flags.
READ = 1 << 0
WRITE = 1 << 1
EXECUTE = 1 << 2
permissions = READ | WRITE
print(bool(permissions & READ))
permissions |= EXECUTE
permissions &= ~WRITE
print(bin(permissions))
```

**Approach:**

- Then trace a tiny example before coding. For this sample, the code demonstrates: Count set bits.
- Code walkthrough:
- `value & (value - 1)` clears the lowest set bit in one operation (Brian Kernighan's algorithm); the loop runs once per set bit.
- Increments `count` each iteration; when `value` reaches `0` all set bits have been counted.
- Prints `4` — `0b101101` has four `1` bits.

**Trade-offs:**
- Fast and compact, but can become unreadable without clear constants and tests.

```python
# Count set bits.
value = 0b101101
count = 0
while value:
    value &= value - 1
    count += 1
print(count)
```
